```python
# alphabet rangoli
def print_rangoli(n):
    for i in range(n):
        s = "-".join(chr(97 + j) for j in range(n - 1, n - i - 2, -1))
        if(n-i<=n-1):
          s += "-" + "-".join(chr(97 + j) for j in range(n - i, n))
        print(s.center(4 * n - 3, "-"))

    for i in range(n - 2, -1, -1):
        s = "-".join(chr(97 + j) for j in range(n - 1, n - i - 2, -1))
        if(n-i<=n-1):
          s += "-" + "-".join(chr(97 + j) for j in range(n - i, n))
        print(s.center(4 * n - 3, "-"))

if __name__ == '__main__':
    n = int(input())
    print_rangoli(n)
```

```python
# swap case
def swap_case(sentence):
    updated_s = ""
    for c in sentence:
        if c.isupper():
            updated_s += c.lower()
        elif c.islower():
            updated_s += c.upper()
        else:
            updated_s += c
    return updated_s


if __name__ == '__main__':
    s = input()
    result = swap_case(s)
    print(result)
```

```python
# find a string
def count_substring(string, sub_string):
    count_sub_string = 0
    for i in range(len(string) - len(sub_string) + 1):
        if string[i : i + len(sub_string)] == sub_string:
            count_sub_string += 1
    return count_sub_string

if __name__ == '__main__':
    string = input().strip()
    sub_string = input().strip()

    count = count_substring(string, sub_string)
    print(count)
```

```python
# string formatting
def print_formatted(number):
    # your code goes here
    width = len(bin(number)[2:])
    for i in range(1, number + 1):
        o = oct(i)[2:]
        h = hex(i)[2:]
        h = h.upper()
        b = bin(i)[2:]
        d = str(i)
        print(f"{d:>{width}} {o:>{width}} {h:>{width}} {b:>{width}}")

if __name__ == '__main__':
    n = int(input())
    print_formatted(n)
```
